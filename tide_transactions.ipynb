```python
import pandas as pd
from quickbooks_sync import get_qb_client
from quickbooks.objects.account import Account
```

```python
df = pd.read_csv("data/tide_transactions.csv")
df.head()
```

```python
summary = df.groupby('Bank description').agg(
    count=('Bank description', 'size'),
    most_common_category=('Match/Categorise', lambda x: x.mode()[0] if not x.mode().empty else None)
).sort_values('count', ascending=False)

summary.head()
```

```python
summary.to_excel('data/bank_transaction_summary.xlsx')
```
